# Data Cleaning

Dataset: **Bank Marketing**

This notebook corresponds to **Section 3.2 "Data Cleaning"** of the paper.
It checks the raw dataset for duplicates, missing values, and inconsistencies,
and produces a cleaned intermediate artifact `bank-cleaned.csv`.

The cleaning step is conservative: only verification checks and the binary
mapping of `default`, `housing`, `loan`, and the target `y` are applied here.
Feature transformations (month conversion, one-hot encoding, scaling) are
performed in `03_data_preprocessing.ipynb`.


```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Paths
ROOT = Path.cwd().resolve()                 # BankMarketing/notebooks
DATA_RAW = ROOT.parent / "data" / "raw"
DATA_PROCESSED = ROOT.parent / "data" / "processed"
DATA_PROCESSED.mkdir(parents=True, exist_ok=True)
```

```python
raw_data = pd.read_csv(DATA_RAW / "bank-full.csv", sep=";")
print("Shape:", raw_data.shape)
```

```python
duplicates = raw_data.duplicated().sum()
print(f"\nNumber of duplicates: {duplicates}")
```

```python
levels = raw_data.apply(pd.Series.unique)
num_levels = raw_data.apply(pd.Series.nunique)
print("Unique values per column:")
print(levels)
print("\nNumber of unique values per column:")
print(num_levels)
```

### Data cleaning


```python
clean_data = raw_data.copy()
```

```python
# Binary column conversion
clean_data['default'] = clean_data['default'].map({'no': 0, 'yes': 1})
clean_data['housing'] = clean_data['housing'].map({'no': 0, 'yes': 1})
clean_data['loan'] = clean_data['loan'].map({'no': 0, 'yes': 1})
clean_data['y'] = clean_data['y'].map({'no': 0, 'yes': 1})
```

```python
cleaned_path = DATA_PROCESSED / "bank-cleaned.csv"
clean_data.to_csv(cleaned_path, index=False)
print(f"Cleaned dataset saved to: {cleaned_path}")
```
